### 1. Data Preprocessing

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Construct the direct download URL for the Google Sheet
csv_url = 'https://docs.google.com/spreadsheets/d/1mnoUgK8YxInzoDQ7P7iBM1HeZ8tcnUSvMU_H1OWndFA/export?format=csv&gid=0'

# Load the dataset
df = pd.read_csv(csv_url)

print("Dataset loaded successfully:")
display(df.head())
print("\nDataset Info:")
display(df.info())
```

### Separate features (X) and target (y)

```python
# Separate input features (X) and target variable (y)
X = df.drop(['Play Tennis', 'No'], axis=1)
y = df['Play Tennis']

print("Features (X) head:")
display(X.head())
print("\nTarget (y) head:")
display(y.head())
```

### Encode categorical features and target variable

```python
# Apply LabelEncoder to all categorical features in X
X_encoded = X.copy()
for column in X.columns:
    le = LabelEncoder()
    X_encoded[column] = le.fit_transform(X_encoded[column])

# Apply LabelEncoder to the target variable y
le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)

print("Encoded Features (X_encoded) head:")
display(X_encoded.head())
print("\nEncoded Target (y_encoded) head:")
display(pd.Series(y_encoded).head())
print("\nOriginal target labels:", le_y.classes_)
```

## 3. Naive Bayes Model Training & Evaluation

```python
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Initialize and train the Categorical Naive Bayes model
cnb = CategoricalNB()
cnb.fit(X_train, y_train)

print("Categorical Naive Bayes model trained successfully.")

# Predict the class labels for the test dataset
y_pred_cnb = cnb.predict(X_test)

# Calculate and display the overall Model Accuracy
accuracy_cnb = accuracy_score(y_test, y_pred_cnb)
print(f"\nCategorical Naive Bayes Model Accuracy: {accuracy_cnb:.4f}")

# Display the Confusion Matrix
conf_matrix_cnb = confusion_matrix(y_test, y_pred_cnb)
print("\nConfusion Matrix for Categorical Naive Bayes:")
display(pd.DataFrame(conf_matrix_cnb, index=le_y.classes_, columns=le_y.classes_))

# Display the Classification Report (Precision, Recall, F1-Score)
class_report_cnb = classification_report(y_test, y_pred_cnb, target_names=le_y.classes_)
print("\nClassification Report for Categorical Naive Bayes:")
print(class_report_cnb)
```

## 4. Single-Sample Inference

```python
from sklearn.preprocessing import LabelEncoder

# Define the specific weather conditions for inference
single_sample_data = {
    'Outlook': ['Sunny'],
    'Temperature': ['Cool'],
    'Humidity': ['High'],
    'Wind': ['Strong']
}
single_sample_df = pd.DataFrame(single_sample_data)

print("Single sample for inference:")
display(single_sample_df)

# To encode the single sample, we need to re-use the LabelEncoders fitted on the training data.
# However, the original code creates new LabelEncoders for each column in X_encoded.
# We need to store and reuse the LabelEncoders for consistent encoding.

# Re-fit LabelEncoders and store them in a dictionary for future use (if not already done)
# This block ensures we have the encoders available if the previous cell was re-run or skipped
le_dict = {}
X_encoded_check = X.copy() # Use original X to fit encoders
for column in X.columns:
    le = LabelEncoder()
    le.fit(X[column]) # Fit on original, unencoded column data
    le_dict[column] = le
    X_encoded_check[column] = le.transform(X[column]) # This is just to ensure le_dict is populated correctly

# Encode the single sample using the stored LabelEncoders
single_sample_encoded = single_sample_df.copy()
for column in single_sample_df.columns:
    if column in le_dict:
        # Ensure the label is known to the encoder, handle unseen labels if necessary
        # For this specific problem, we assume the labels are seen.
        single_sample_encoded[column] = le_dict[column].transform(single_sample_encoded[column])
    else:
        print(f"Warning: Encoder not found for column {column}. Skipping encoding for this column.")

print("\nEncoded single sample:")
display(single_sample_encoded)

# Predict the class label for the single sample
predicted_label_encoded = cnb.predict(single_sample_encoded)
predicted_label_original = le_y.inverse_transform(predicted_label_encoded)

# Predict the class probabilities for the single sample
predicted_probabilities = cnb.predict_proba(single_sample_encoded)

print(f"\nPredicted class label: {predicted_label_original[0]}")
print("Predicted class probabilities:")
probability_df = pd.DataFrame(predicted_probabilities, columns=le_y.classes_)
display(probability_df)
```

## 5. Model Comparison

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import pandas as pd

# Dictionary to store models and their results
models = {
    'Categorical Naive Bayes': cnb, # Reuse the already trained CNB model
}

accuracies = {}
single_sample_predictions = {}
single_sample_probabilities = {}

# --- Train and Evaluate Decision Tree Classifier ---
dt_classifier = DecisionTreeClassifier(random_state=42)
dt_classifier.fit(X_train, y_train)
y_pred_dt = dt_classifier.predict(X_test)
accuracies['Decision Tree'] = accuracy_score(y_test, y_pred_dt)
single_sample_predictions['Decision Tree'] = le_y.inverse_transform(dt_classifier.predict(single_sample_encoded))[0]
single_sample_probabilities['Decision Tree'] = dt_classifier.predict_proba(single_sample_encoded)[0]

# --- Train and Evaluate Logistic Regression Classifier ---
# Logistic Regression requires features to be non-negative for some solvers and can handle categorical data encoded numerically
# For simplicity, we'll use default solver or 'liblinear' for smaller datasets.
# C is inverse of regularization strength; smaller values specify stronger regularization.
lr_classifier = LogisticRegression(random_state=42, solver='liblinear')
lr_classifier.fit(X_train, y_train)
y_pred_lr = lr_classifier.predict(X_test)
accuracies['Logistic Regression'] = accuracy_score(y_test, y_pred_lr)
single_sample_predictions['Logistic Regression'] = le_y.inverse_transform(lr_classifier.predict(single_sample_encoded))[0]
single_sample_probabilities['Logistic Regression'] = lr_classifier.predict_proba(single_sample_encoded)[0]

# --- Train and Evaluate SVM Classifier ---
# SVM (SVC) does not natively support probability prediction with predict_proba unless enabled.
# We set probability=True to get probabilities.
svm_classifier = SVC(random_state=42, probability=True)
svm_classifier.fit(X_train, y_train)
y_pred_svm = svm_classifier.predict(X_test)
accuracies['SVM'] = accuracy_score(y_test, y_pred_svm)
single_sample_predictions['SVM'] = le_y.inverse_transform(svm_classifier.predict(single_sample_encoded))[0]
single_sample_probabilities['SVM'] = svm_classifier.predict_proba(single_sample_encoded)[0]


# --- Collect results for Naive Bayes ---
accuracies['Categorical Naive Bayes'] = accuracy_cnb # From previous step
single_sample_predictions['Categorical Naive Bayes'] = predicted_label_original[0]
single_sample_probabilities['Categorical Naive Bayes'] = predicted_probabilities[0]

# --- Display Comparison Table ---
comparison_data = []
for model_name in ['Categorical Naive Bayes', 'Decision Tree', 'Logistic Regression', 'SVM']:
    probabilities_str = ', '.join([f'{le_y.classes_[i]}: {p:.2f}' for i, p in enumerate(single_sample_probabilities[model_name])])
    comparison_data.append({
        'Model': model_name,
        'Test Accuracy': f"{accuracies[model_name]:.4f}",
        'Single Sample Prediction': single_sample_predictions[model_name],
        'Single Sample Probabilities': probabilities_str
    })

comparison_df = pd.DataFrame(comparison_data)
print("\n--- Model Comparison ---")
display(comparison_df)

# --- Analysis Report ---
print("\n## Analysis Report")
print("The models produced different predictions and probability scores due to their underlying algorithms and assumptions. Categorical Naive Bayes assumes feature independence, which might not hold perfectly in real-world data, but can be surprisingly robust. Decision Trees make decisions based on feature thresholds, which can lead to clear, but sometimes overfitted, boundaries. Logistic Regression uses a linear combination of features to predict probability, while SVM aims to find the optimal hyperplane to separate classes, often focusing on margin maximization rather than probability estimation (though it can provide them). These fundamental differences lead to varied performance and predictions on specific instances, especially with limited data or complex relationships.")
```
