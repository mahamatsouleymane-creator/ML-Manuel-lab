### Step 1: Import Required Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score
from sklearn.metrics import confusion_matrix
```

### Step 2: Download and Load the Breast Cancer Wisconsin Dataset

```python
# Load Dataset

cancer = load_breast_cancer()

X = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y = pd.Series(cancer.target, name="Target")

# Combine into one DataFrame
df = X.copy()
df["Target"] = y

print("Dataset Loaded Successfully")
display(df.head())
```

### Step 3: Perform Exploratory Data Analysis (EDA)

#### Dataset Shape

```python
print("Shape of Dataset:", df.shape)
```

#### Dataset Information

```python
df.info()
```

#### Statistical Summary

```python
display(df.describe())
```

#### Target Class Distribution

```python
print(df["Target"].value_counts())
```

#### Plot Target Distribution

```python
plt.figure(figsize=(5,4))
sns.countplot(x="Target", data=df)

plt.title("Target Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks([0,1],["Malignant","Benign"])
plt.show()
```

#### Correlation Heatmap

```python
plt.figure(figsize=(12,10))

sns.heatmap(df.corr(), cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

### Step 4: Check Missing Values and Data Preprocessing

#### Missing Values

```python
print(df.isnull().sum())
```

#### Total Missing Values

Output should be

#### Separate Features and Target

```python
X = df.drop("Target", axis=1)

y = df["Target"]
```

#### Standardize Features

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

### Step 5: Split Dataset into Training and Testing Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Shape :", X_train.shape)
print("Testing Shape  :", X_test.shape)
```

### Step 6: Train Logistic Regression Classifier

```python
lr = LogisticRegression(max_iter=1000)

lr.fit(X_train, y_train)

lr_pred = lr.predict(X_test)

print("Logistic Regression Model Trained Successfully")
```

### Step 7: Train K-Nearest Neighbors (KNN) Classifier

```python
knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train, y_train)

knn_pred = knn.predict(X_test)

print("KNN Model Trained Successfully")
```

### Step 8: Evaluate Logistic Regression

```python
lr_accuracy = accuracy_score(y_test, lr_pred)
lr_precision = precision_score(y_test, lr_pred)
lr_recall = recall_score(y_test, lr_pred)
lr_f1 = f1_score(y_test, lr_pred)
lr_cm = confusion_matrix(y_test, lr_pred)

print("Logistic Regression Performance")
print("-------------------------------")
print("Accuracy :", lr_accuracy)
print("Precision:", lr_precision)
print("Recall   :", lr_recall)
print("F1 Score :", lr_f1)

print("\nConfusion Matrix")
print(lr_cm)
```

#### Logistic Regression Confusion Matrix Graph

```python
plt.figure(figsize=(5,4))

sns.heatmap(
    lr_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Malignant","Benign"],
    yticklabels=["Malignant","Benign"]
)

plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

### Step 9: Evaluate KNN

```python
knn_accuracy = accuracy_score(y_test, knn_pred)
knn_precision = precision_score(y_test, knn_pred)
knn_recall = recall_score(y_test, knn_pred)
knn_f1 = f1_score(y_test, knn_pred)
knn_cm = confusion_matrix(y_test, knn_pred)

print("KNN Performance")
print("-------------------------------")
print("Accuracy :", knn_accuracy)
print("Precision:", knn_precision)
print("Recall   :", knn_recall)
print("F1 Score :", knn_f1)

print("\nConfusion Matrix")
print(knn_cm)
```

#### KNN Confusion Matrix Graph

```python
plt.figure(figsize=(5,4))

sns.heatmap(
    knn_cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Malignant","Benign"],
    yticklabels=["Malignant","Benign"]
)

plt.title("KNN Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

### Step 10: Compare Both Models

```python
comparison = pd.DataFrame({
    "Metric":["Accuracy","Precision","Recall","F1 Score"],
    "Logistic Regression":[
        lr_accuracy,
        lr_precision,
        lr_recall,
        lr_f1
    ],
    "KNN":[
        knn_accuracy,
        knn_precision,
        knn_recall,
        knn_f1
    ]
})
```

#### Print Comparison Table

```python
print(comparison)
```

```python
comparison.set_index("Metric").plot(kind="bar", figsize=(8,5))

plt.title("Logistic Regression vs KNN")
plt.xlabel("Evaluation Metrics")
plt.ylabel("Score")
plt.ylim(0.90,1.00)
plt.grid(axis="y")

plt.show()
```

### Step 12: Interpretation

```python
print("INTERPRETATION")
print("="*60)

if lr_accuracy > knn_accuracy:
    print("Logistic Regression achieved better performance than KNN.")
    print("It has higher Accuracy, Precision, Recall and F1 Score.")
    print("It also produced fewer misclassifications in the Confusion Matrix.")
    print("Therefore, Logistic Regression is the better classifier for the Breast Cancer Wisconsin dataset.")
else:
    print("KNN performed better than Logistic Regression.")
```
