```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

aqi.columns = aqi.columns.str.strip()

aqi['AQI'] = pd.to_numeric(
    aqi['AQI'],
    errors='coerce'
)

aqi['AQI'].fillna(
    aqi['AQI'].median(),
    inplace=True
)

plt.figure(figsize=(10,5))

sns.histplot(
    data=aqi,
    x='AQI',
    bins=40,
    kde=True
)

plt.title("AQI Distribution")
plt.xlabel("AQI")
plt.ylabel("Frequency")

plt.show()
```

TASK 5
Cell 9: Outlier Detection

```python
Q1 = aqi['AQI'].quantile(0.25)

Q3 = aqi['AQI'].quantile(0.75)

IQR = Q3-Q1

lower = Q1-(1.5*IQR)

upper = Q3+(1.5*IQR)

print("Lower:", lower)

print("Upper:", upper)

outliers = aqi[
    (aqi['AQI'] < lower) |
    (aqi['AQI'] > upper)
]

print("Outliers:", len(outliers))
```

TASK 6
Cell 13: Year Trend

```python
aqi['Date'] = pd.to_datetime(
    aqi['Date']
)

aqi['Year'] = aqi['Date'].dt.year

yearly = aqi.groupby(
    'Year'
)['AQI'].mean().reset_index()

yearly
```

Cell 14: Trend Plot

```python
plt.figure(figsize=(10,5))

sns.lineplot(
    data=yearly,
    x='Year',
    y='AQI',
    marker='o'
)

plt.title("Average AQI Over Time")

plt.ylabel("Average AQI")

plt.show()
```

Most & Least Polluted Year

```python
print(
    yearly.loc[
        yearly['AQI'].idxmax()
    ]
)

print(
    yearly.loc[
        yearly['AQI'].idxmin()
    ]
)
```
